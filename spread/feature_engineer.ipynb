```python
df_org = pd.read_csv('spread.csv').set_index('datetime', drop=True)
print(df_org.keys())
df_org
```

```python
df = df_org.copy()
df.index = pd.to_datetime(df.index, utc=True)
df
```

```python
df_feature = pd.DataFrame(index=df_org.index)
df_feature
```

```python
df_feature['month_sin'] = np.sin(2 * np.pi * (df.index.month - 1) / 12)
df_feature['dayofweek_sin'] = np.sin(2 * np.pi * df.index.dayofweek / 7)
df_feature['dayofmonth_sin'] = np.sin(2 * np.pi * (df.index.day - 1) / 31)
df_feature
```

```python
columns_for_iqr = ['da_price', 'rt_price', 'Biomass', 'Coal', 'Gas', 'Gas-CC',
       'Hydro', 'Nuclear', 'Other', 'Solar', 'Wind']
columns_to_scale = [c for c in df.columns if c in columns_for_iqr]
print("Columns to scale:", columns_to_scale)

scaler = RobustScaler()
df_feature[columns_to_scale] = scaler.fit_transform(df[columns_to_scale])
df_feature[columns_to_scale] = df_feature[columns_to_scale].shift(24)
df_feature
```

```python
df_feature['spread'] = df['spread']
df_feature.to_csv('spread_features.csv')
df_feature
```
